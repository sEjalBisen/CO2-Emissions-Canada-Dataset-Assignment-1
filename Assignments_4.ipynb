````python
# Run this code cell
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

url = 'https://new-assets.ccbp.in/frontend/content/aiml/classical-ml/co2_emissions_canada.csv'

# Upload the data on colab and execute this code cell
data = pd.read_csv(url)
data.head()
````

---
**Q1. How much CO2 is emitted per liter of fuel consumed for each car model?
Hint: (CO2 Emissions / Fuel Consumption).**

````python
# Write your code here
data["CO2 per Liter"] = data["CO2 Emissions(g/km)"] / data["Fuel Consumption Comb (L/100 km)"]
print(data[["Make", "Model", "CO2 per Liter"]].head(5))

````

---
Q2. Compute the correlation matrix to analyze the relationships between features such as engine size, fuel consumption, and CO₂ emissions in the given order.
```
Engine Size(L)
Cylinders
Fuel Consumption City (L/100 km)
Fuel Consumption Hwy (L/100 km)
Fuel Consumption Comb (L/100 km)
Fuel Consumption Comb (mpg)
CO2 Emissions(g/km)
```

````python
# If the output is not displayed as expected, its not code issue, but with how it is displayed, try below code lines
# Ensure full output display for large DataFrames
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Select only numerical columns for correlation analysis
numeric_cols = data[['Engine Size(L)', 'Cylinders',
                     'Fuel Consumption City (L/100 km)',
                     'Fuel Consumption Hwy (L/100 km)',
                     'Fuel Consumption Comb (L/100 km)',
                     'Fuel Consumption Comb (mpg)',
                     'CO2 Emissions(g/km)']]

# Compute correlation matrix
corr_matrix = numeric_cols.corr()

# Print full correlation matrix
print(corr_matrix)
````

````python
# Write your code here to create a visualization
plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Vehicle Attributes')
plt.tight_layout()
plt.show()
````

---
Q3. What is the correlation between engine size and CO₂ emissions?

````python
# write your code here
# Filter the data for vehicles with Engine Size > 2.0L
filtered_data = data[data['Engine Size(L)'] > 2.0]

# Compute correlation between Engine Size and CO2 Emissions
corr_data = filtered_data[['Engine Size(L)', 'CO2 Emissions(g/km)']].corr()
print(corr_data)
````

````python
# Write your code here to create a visualization
plt.figure()
plt.scatter(data['Engine Size(L)'], data['CO2 Emissions(g/km)'])
plt.xlabel('Engine Size (L)')
plt.ylabel('CO2 Emissions (g/km)')
plt.title('Engine Size vs CO2 Emissions')
plt.show()

````

---
**Q4. What is the average CO₂ emission for each vehicle class?**

````python
# write your code here
table1 = (
    data.groupby('Vehicle Class')
      .agg(avg_CO2=('CO2 Emissions(g/km)','mean'))
      .sort_values('avg_CO2', ascending=False)
      .reset_index()
)
print(table1)
````

````python
# Write your code here to create a visualization
plt.figure()
plt.bar(table1['Vehicle Class'], table1['avg_CO2'])
plt.title('Average CO₂ Emissions by Vehicle Class')
plt.xlabel('Vehicle Class')
plt.ylabel('Avg CO₂ (g/km)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

````

---
**Q5. Generate a summary table showing the minimum, maximum, and average CO₂ emissions for each fuel type.**

````python
# Group by 'Fuel Type' and compute summary statistics
co2_summary = data.groupby('Fuel Type')['CO2 Emissions(g/km)'].agg(['min', 'max', 'mean']).reset_index()
co2_summary.columns = ['Fuel Type', 'Min CO₂ Emissions', 'Max CO₂ Emissions', 'Avg CO₂ Emissions']

print(co2_summary)

````

````python
# Write your code here to create a visualization
plt.figure(figsize=(10, 6))
sns.violinplot(x='Fuel Type', y='CO2 Emissions(g/km)', data=data)
plt.title('Distribution of CO₂ Emissions by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('CO₂ Emissions (g/km)')
plt.tight_layout()
plt.show()

````

**Q6. Create a table showing the average CO₂ emissions for each engine size and fuel type combination.**

````python
# Create summary table: average CO₂ emissions by engine size and fuel type
emission_summary = data.groupby(['Engine Size(L)', 'Fuel Type'])['CO2 Emissions(g/km)'].mean().reset_index()
emission_summary.columns = ['Engine Size (L)', 'Fuel Type', 'Avg CO₂ Emissions (g/km)']

print(emission_summary.head(10))  # Display first 10 rows
````

````python
# Write your code here to create a visualization
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Engine Size(L)', y='CO2 Emissions(g/km)', data=data, hue='Fuel Type')
plt.title('Engine Size vs CO₂ Emissions')
plt.xlabel('Engine Size (L)')
plt.ylabel('CO₂ Emissions (g/km)')
plt.tight_layout()
plt.show()
````

---
**Q7. Display a table showing the min, max, median, and average CO₂ emissions for each vehicle class.**

````python
# Write your code here
# Summary stats by vehicle class
vehicle_class_summary = data.groupby('Vehicle Class')['CO2 Emissions(g/km)'].agg(['min', 'max', 'median', 'mean']).reset_index()
vehicle_class_summary.columns = ['Vehicle Class', 'Min CO₂', 'Max CO₂', 'Median CO₂', 'Avg CO₂']

print(vehicle_class_summary.head(10))
````

````python
# Write your code here to create a visualization
plt.figure(figsize=(12, 6))
sns.boxplot(x='Vehicle Class', y='CO2 Emissions(g/km)', data=data)
plt.title('CO₂ Emissions by Vehicle Class')
plt.xlabel('Vehicle Class')
plt.ylabel('CO₂ Emissions (g/km)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````
